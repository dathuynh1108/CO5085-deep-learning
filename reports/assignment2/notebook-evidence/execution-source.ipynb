# A2 · Image segmentation with boundary-guided feature fusion

Does a boundary-dependent fusion weight improve segmentation of the transition region in Oxford-IIIT Pet trimaps? Both variants use the same ResNet18 encoder, data, loss function and training budget. The feature fusion rule is the only difference between them.

The fusion mechanism is adapted from Bag in *PIDNet: A Real-time Semantic Segmentation Network Inspired by PID Controllers* by Xu, Xiong and Bhattacharyya, IEEE/CVF CVPR 2023, pp. 19529–19539. [Paper and DOI](https://doi.org/10.1109/CVPR52729.2023.01871)

The scope of AI assistance is documented in [AI_USE_REPORT.md](../AI_USE_REPORT.md).

## 1. Experimental protocol

The experiment uses the full Oxford-IIIT Pet dataset, with 3,310 training images, 370 validation images and 3,669 test images. Each of the two variants is trained for 12 epochs with seeds 42, 43 and 44. The shared protocol uses batch size 32, AdamW and cosine learning-rate decay. Both models start from ImageNet-pretrained ResNet18 weights with identical initialization for each paired seed.

GPU execution uses BF16 autocast, with the loss computed in FP32. These settings are held constant between the variants.

## 3. Model, training and validation

Given detail features $D$ and context features $C$, both variants compute $F = gD+(1-g)C$. The weight $g$ is $0.5$ for fixed fusion and $\sigma(e)$ for boundary-guided fusion, where $e$ is the boundary head logit. Both models minimize the sum of segmentation cross-entropy and balanced boundary BCE, with a coefficient of $0.25$ for BCE. Shared boundary supervision isolates the effect of the fusion rule in this setting.

Each epoch consists of AdamW training updates followed by validation. Cosine annealing adjusts the learning rates of the encoder and prediction heads. The checkpoint is updated only when validation mIoU improves, so ties retain the earliest epoch.

```python
def train_one(variant, seed, path, identity, train_set, val_set):
    path.mkdir(parents=True, exist_ok=False)
    seed_everything(seed)
    model = PetSegmenter(variant, pretrained=True)
    initial_state = state_fingerprint(model.state_dict())
    model.to(device=device, memory_format=torch.channels_last)
    head_params = [
        param
        for name, param in model.named_parameters()
        if not name.startswith("encoder.")
    ]
    optimizer = torch.optim.AdamW(
        [
            {"params": model.encoder.parameters(), "lr": config["encoder_lr"]},
            {"params": head_params, "lr": config["head_lr"]},
        ],
        weight_decay=config["weight_decay"],
    )
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=config["epochs"]
    )
    train_batches = loader(train_set, config, workers, device, seed, shuffle=True)
    val_batches = loader(val_set, config, workers, device, seed)
    metadata = dict(
        identity,
        config=config,
        variant=variant,
        seed=seed,
        workers=workers,
        initial_state_sha256=initial_state,
        params=sum(p.numel() for p in model.parameters()),
    )
    atomic_json(path / "protocol.json", metadata)

    history = []
    best_score = -1.0
    if device.type == "cuda":
        torch.cuda.reset_peak_memory_stats(device)
    for epoch in range(1, config["epochs"] + 1):
        model.train()
        synchronize(device)
        started = time.perf_counter()
        loss_sum = torch.zeros((), device=device)
        for images, masks in train_batches:
            masks = masks.to(device, non_blocking=True, dtype=torch.long)
            optimizer.zero_grad(set_to_none=True)
            with torch.autocast(
                device.type, dtype=torch.bfloat16, enabled=device.type == "cuda"
            ):
                logits, edges = model(normalize(images, device))
            loss = segmentation_loss(logits, edges, masks, config["boundary_weight"])
            loss.backward()
            optimizer.step()
            loss_sum += loss.detach() * len(images)

        synchronize(device)
        seconds = time.perf_counter() - started
        train_loss = float(loss_sum / len(train_set))
        validation, _ = evaluate(model, val_batches, device)
        if not np.isfinite([train_loss, validation["miou"]]).all():
            raise RuntimeError(f"Non-finite metric at {path}, epoch {epoch}.")
        if validation["miou"] > best_score:
            best_score, best_epoch = validation["miou"], epoch
            atomic_torch_save(path / "best.pt", model.state_dict())
        history.append(
            {
                "epoch": epoch,
                "train_loss": train_loss,
                "train_seconds": seconds,
                "val_miou": validation["miou"],
                "val_macro_dice": validation["macro_dice"],
                "val_transition_iou": validation["iou"][2],
                "encoder_lr": optimizer.param_groups[0]["lr"],
                "head_lr": optimizer.param_groups[1]["lr"],
            }
        )
        atomic_csv(path / "history.csv", history)
        print(
            f"{variant:6} seed={seed} epoch={epoch:02d} loss={train_loss:.4f} "
            f"val_mIoU={validation['miou']:.4f} train={seconds:.1f}s",
            flush=True,
        )
        scheduler.step()
    atomic_json(
        path / "run.json",
        dict(
            metadata,
            epochs=len(history),
            best_epoch=best_epoch,
            best_val_miou=best_score,
            total_train_seconds=sum(row["train_seconds"] for row in history),
            peak_allocated_mib=(
                torch.cuda.max_memory_allocated(device) / 2**20
                if device.type == "cuda"
                else None
            ),
            checkpoint_sha256=sha256(path / "best.pt"),
            history_sha256=sha256(path / "history.csv"),
        ),
    )
```

```python
def selected_indices(partition):
    total = manifest["splits"][partition]["size"]
    if not smoke:
        return range(total)
    limits = {"train": 32, "validation": 16, "test": 16}
    rng = np.random.default_rng(config["split_seed"])
    return np.sort(
        rng.choice(total, size=min(limits[partition], total), replace=False)
    ).tolist()


def subset_for_run(dataset, partition):
    return Subset(dataset, selected_indices(partition)) if smoke else dataset
```

```python
identity = dict(
    source_identity,
    notebook_config_sha256=json_hash(config),
    smoke=smoke,
    train_loop_sha256=json_hash({"source": inspect.getsource(train_one)}),
    sample_indices_sha256=json_hash(
        {part: list(selected_indices(part)) for part in manifest["splits"]}
    ),
    runtime=environment_info(device)
    | {
        "torchvision": torchvision.__version__,
        "numpy": np.__version__,
        "pillow": PIL.__version__,
        "precision": precision,
        "workers": workers,
    },
)
```

```python
train_set = subset_for_run(train_full, "train")
val_set = subset_for_run(validation_full, "validation")
lock_path = output / "frozen.json"
for variant, seed, run_path in run_paths(output, config):
    if (run_path / "run.json").is_file():
        record = json.loads((run_path / "run.json").read_text())
        if (
            any(record.get(key) != value for key, value in identity.items())
            or (record["variant"], record["seed"], record["epochs"])
            != (variant, seed, config["epochs"])
            or record["checkpoint_sha256"] != sha256(run_path / "best.pt")
            or record["history_sha256"] != sha256(run_path / "history.csv")
        ):
            raise ValueError(
                f"Run evidence does not match the notebook protocol: {run_path}"
            )
        print(f"Verified completed run: {variant}, seed {seed}")
        continue
    if lock_path.exists() or run_path.exists():
        raise FileExistsError(
            f"An incomplete run or frozen protocol exists: {run_path}. Choose a new CO5085_OUTPUT_ROOT."
        )
    train_one(variant, seed, run_path, identity, train_set, val_set)
```

```python
results = {}
summary_rows = []
for variant, seed, run_path in run_paths(output, config):
    result = json.loads((run_path / "test.json").read_text())
    results[variant, seed] = result
    summary_rows.append(
        dict(
            variant=variant,
            seed=seed,
            samples=result["samples"],
            miou=result["miou"],
            macro_dice=result["macro_dice"],
            transition_iou=result["iou"][2],
        )
    )
atomic_csv(output / "metrics_by_seed.csv", summary_rows)
```

```python
metrics = ("miou", "macro_dice", "transition_iou")
lines = [
    "| Variant | Seed | mIoU (%) | Macro-Dice (%) | Transition IoU (%) |",
    "|---|---:|---:|---:|---:|",
]
for row in summary_rows:
    values = " | ".join(f"{100 * row[name]:.2f}" for name in metrics)
    lines.append(f"| {row['variant']} | {row['seed']} | {values} |")
```

```python
aggregate = {}
for variant in config["variants"]:
    aggregate[variant] = {}
    for metric in metrics:
        values = [row[metric] for row in summary_rows if row["variant"] == variant]
        aggregate[variant][metric] = dict(
            mean=mean(values), sample_sd=stdev(values) if len(values) > 1 else None
        )
    values = [
        f"{100 * item['mean']:.2f}"
        + (f" ± {100 * item['sample_sd']:.2f}" if item["sample_sd"] is not None else "")
        for item in aggregate[variant].values()
    ]
    lines.append(f"| {variant} | Aggregate | {' | '.join(values)} |")
```

## 6. Scope of the comparison with PIDNet

This experiment isolates the effect of boundary-guided fusion weights in a ResNet18 segmenter. It adapts the Bag fusion rule rather than reproducing the complete PIDNet architecture, including its independent D branch, Pag and PAPPM modules. Both local variants receive the same boundary supervision, so their comparison measures the additional effect of using the predicted boundary signal in fusion.

The mIoU differences are interpreted within the Oxford-IIIT Pet setting, alongside transition-region IoU and the rates at which this region is confused with the pet or background. The trimap transition region has a finite width, so results depend on annotation and image resolution. A result on this dataset does not establish the same effect on Cityscapes or CamVid, where PIDNet was evaluated under different architectures and protocols.

The [A2 report](../reports/assignment2/assignment2.pdf) analyzes the original paper's method, experiments and limitations, and compares its published ablations with this study.
